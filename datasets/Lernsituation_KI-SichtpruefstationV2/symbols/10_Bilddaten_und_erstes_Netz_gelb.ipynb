### [Aufgabe C2] Das erste Netz bauen
So soll das Netz aussehen (wie in der Grafik oben):

| Schicht | Aufgabe |
|---|---|
| `keras.Input(shape=(32, 32))` | Eingang: ein Bild mit 32 × 32 Pixeln |
| `layers.Flatten()` | Matrix → Vektor mit 1024 Zahlen |
| `layers.Dense(32, activation="relu")` | Schicht mit 32 Neuronen |
| `layers.Dropout(0.2)` | schaltet beim Lernen zufällig 20 % der Neuronen ab |
| `layers.Dense(128, activation="relu")` | Schicht mit 128 Neuronen |
| `layers.Dropout(0.2)` | |
| `layers.Dense(64, activation="relu")` | Schicht mit 64 Neuronen |
| `layers.Dropout(0.2)` | |
| `layers.Dense(16, activation="relu")` | Schicht mit 16 Neuronen |
| `layers.Dropout(0.2)` | |
| `layers.Dense(ANZAHL_KLASSEN, activation="softmax")` | Ausgabe: eine Wahrscheinlichkeit je Klasse |

**Dropout** verhindert, dass das Netz die Trainingsbilder einfach auswendig lernt. Beim Erkennen (nach dem Training) ist Dropout automatisch ausgeschaltet.

**Befehle:** `layers.Dense(anzahl_neuronen, activation="relu"),` und `layers.Dropout(0.2),` – das Komma am Ende jeder Zeile nicht vergessen.

````python
modell = keras.Sequential([
    keras.Input(shape=(32, 32)),
    layers.Flatten(),
    # Schicht mit 32 Neuronen, Aktivierung "relu"
    # Dropout 0.2
    # Schicht mit 128 Neuronen, Aktivierung "relu"
    # Dropout 0.2
    # Schicht mit 64 Neuronen, Aktivierung "relu"
    # Dropout 0.2
    # Schicht mit 16 Neuronen, Aktivierung "relu"
    # Dropout 0.2
    # Ausgabeschicht: ANZAHL_KLASSEN Neuronen, Aktivierung "softmax"
])
modell.summary()

kh.pruefe_gleich(modell.output_shape, (None, ANZAHL_KLASSEN), "Ausgabe des Netzes")
kh.pruefe_gleich(modell.count_params(), 46388, "Anzahl Gewichte")
````
